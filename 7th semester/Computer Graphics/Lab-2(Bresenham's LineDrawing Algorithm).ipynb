## Bresenham's LineDrawing Algorithm

#### Algorithm:

0. input x1, y1 and x2, y2
1. Calculate dx = x2 - x1
2. Calculate dy = y2 - y1
3. Calculate p = 2dy - dx
4. Set x = x1 & y = y1
5. Repeat While x <= x2:

           plot(x, y)
    
           if p >= 0:
               y = y + 1
               p = p + 2(dy - dx)
           else:
               p = p + 2dy
    
           x = x + 1
6. Stop
      

#### Code:

```python
import matplotlib.pyplot as plt
```

```python
x1, y1 = map(int, input("Enter the x1, y1 number: ").split())
x2, y2 = map(int, input("Enter the x2, y2 number: ").split())
dx = x2 - x1
dy = y2 - y1
p = 2* dy - dx
x, y = x1, y1
x_val, y_val = [], []
while x <= x2:
    x_val += [x]
    y_val += [y]
    if p >= 0:
        y += 1
        p += 2 * (dy - dx)
    else:p += 2 * dy
    x += 1

```

```python
# draw the line
plt.plot(x_val, y_val, 'o-',color = 'green', label = 'Bressenham')
plt.title("Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```
