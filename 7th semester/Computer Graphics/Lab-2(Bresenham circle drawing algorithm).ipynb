 <h1>Bresenham Circle Drawing Algorithm </h1>

### **Algorithm:**

0. input radious r
1. set x = 0
2. set y = r
3. set p = 3 - 2r
4. Repeat x <= y:
   
           plot(x, y)
           plot(y, x)
           plot(-x, y)
           plot(-y, x)
           plot(x, -y)
           plot(y, -x)
           plot(-x, -y)
           plot(-y, -x)
        
           if p < 0: p = p + 4x + 6
           else:
               p = p + 4(x - y) + 10
               y = y - 1
           x = x + 1
6. stop

#### **Code:**

```python
import matplotlib.pyplot as plt
r = int(input("Enter the value of Radious: "))
x, y = 0, r
p = 3 - 2 * r
x_val, y_val = [],[]

while x <= y:
    x_val += [x]
    y_val += [y]

    x_val += [y]
    y_val += [x]

    x_val += [-x]
    y_val += [y]

    x_val += [-y]
    y_val += [x]

    x_val += [x]
    y_val += [-y]

    x_val += [y]
    y_val += [-x]

    x_val += [-x]
    y_val += [-y]

    x_val += [-y]
    y_val += [-x]

    if p < 0: p += 4 * x + 6
    else: 
        p += 4 * (x - y) + 10
        y -= 1
    x += 1


```

```python
# Drawing

plt.plot(x_val, y_val, '*', color = 'green')
plt.title("Bresenham Circle Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.axis("equal")
plt.show()
```
