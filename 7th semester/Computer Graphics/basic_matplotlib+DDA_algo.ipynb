# DDA(Digital Difference Analyzer)

```python
x1, y1 = 2, 3
x2, y2 = 13, 9
dx, dy = x2-x1, y2-y1
steps = max(abs(dx), abs(dy))
x_inc = dx/steps
y_inc = dy/steps
x, y = x1, y1

x_point, y_point = [],[]
for i in range(steps):
    x_point.append(round(x))
    y_point.append(round(y))

    x = x + x_inc
    y = y + y_inc

plt.scatter(x_point, y_point, label = "DDA", color = "green")
plt.title("DDA Drawing Algorithm")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.plot(x_point, y_point, color = "green")
plt.legend()
plt.grid()
plt.savefig("DDA.jpg")
plt.show()
```
