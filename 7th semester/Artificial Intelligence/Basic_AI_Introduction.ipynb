```python
marks = np.array([120,125,130,131,133,155])
print("Mean:", marks.mean())
print("Max: ", marks.max())
print("Standard Deviation: ", marks.std())
print("Median: ", np.median(marks))
print("Variance: ", marks.var())
print("Sort: ", np.sort(marks))
print("Sum: ", np.sum(marks))
```
